```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Generate synthetic demographic distribution data (Task 01)
np.random.seed(42)
ages = np.random.normal(loc=32, scale=12, size=1000).clip(1, 90).astype(int)
genders = np.random.choice(['Male', 'Female', 'Other'], size=1000, p=[0.49, 0.49, 0.02])

# Plot 1: Age Histogram (Continuous Variable)
plt.figure(figsize=(9, 4.5))
sns.histplot(ages, bins=15, kde=True, color='#2563eb', edgecolor='black')
plt.title('Population Age Distribution', fontsize=13, fontweight='bold')
plt.xlabel('Age (Years)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.savefig('task01_age_histogram.png', dpi=300)
plt.show()

# Plot 2: Gender Bar Chart (Categorical Variable)
plt.figure(figsize=(6, 4))
sns.countplot(x=genders, order=['Male', 'Female', 'Other'], palette=['#3b82f6', '#ec4899', '#10b981'], hue=genders, legend=False)
plt.title('Population Gender Distribution', fontsize=13, fontweight='bold')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.tight_layout()
plt.savefig('task01_gender_barchart.png', dpi=300)
plt.show()
```
